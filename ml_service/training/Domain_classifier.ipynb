```python
import pandas as pd
```

```python
df = pd.read_csv('data/cleaned_query_domain_dataset.csv')
```

```python
df.drop(columns=['topic'], inplace=True)
```

```python
df
```

```python
# label encoding the label column
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(df['label'])
print("classes : ",le.classes_)
```

```python
# tfidf vectorization of the query column
from sklearn.feature_extraction.text import TfidfVectorizer 
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df['query']).toarray()
```

```python
X
```

```python
y
```

```python
# splitting the dataset into train and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Logistic Regression model
from sklearn.linear_model import LogisticRegression
L_model = LogisticRegression()
L_model.fit(X_train, y_train)
y_pred = L_model.predict(X_test)
```

```python
from sklearn.metrics import classification_report, accuracy_score

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Naive Bayes model
from sklearn.naive_bayes import MultinomialNB
NB_model = MultinomialNB()
NB_model.fit(X_train, y_train)
y_pred_nb = NB_model.predict(X_test)
```

```python
from sklearn.metrics import classification_report, accuracy_score

print("Naive Bayes Accuracy:", accuracy_score(y_test, y_pred_nb))
print("Naive Bayes Classification Report:")
print(classification_report(y_test, y_pred_nb))
```

```python
# Exporting the model and vectorizer using joblib
import joblib

joblib.dump(L_model, "domain_classifier.joblib")
joblib.dump(vectorizer, "domain_classifier_vectorizer.joblib")
```
