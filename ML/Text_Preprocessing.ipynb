```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('noisey_query_domain_dataset.csv')
```

```python
df.sample(50)
```

```python
df.isnull().sum()
```

```python
# Lowercase the text
df['query'] = df['query'].str.lower()
```

```python
# Remove HTML tags
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)

df['query'] = df['query'].apply(remove_html_tags)
```

```python
# Remove URLs
def remove_urls(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)

df['query'] = df['query'].apply(remove_urls)
```

```python
# Remove punctuation

exclude = '!"#$%&\'()*,-./:;<=>?@[\\]^_`{|}~'
def remove_punc(text):
    for char in exclude:
        text = text.replace(char,'')
    return text

df['query'] = df['query'].apply(remove_punc)
```

```python
# Remove stopwords

from nltk.corpus import stopwords

def remove_stopwords(text):
    new_text = []
    
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)

df['query'] = df['query'].apply(remove_stopwords)
```

```python
# Remove emojis
import re
def remove_emoji(text):
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags
        u"\U00002700-\U000027BF"  # dingbats
        u"\U000024C2-\U0001F251"
        u"\U0001F900-\U0001F9FF"  # supplemental symbols & pictographs
        u"\U0001FA70-\U0001FAFF"  # extended symbols
        u"\U0001F018-\U0001F270"  # various additional symbols
        u"\U0001F000-\U0001F02F"  # Mahjong tiles
        "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)

df['query'] = df['query'].apply(remove_emoji)
```

```python
df.to_csv('cleaned_query_domain_dataset.csv', index=False)
```
